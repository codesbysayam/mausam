## Notebook 03: NWP Multi-Model Comparison & Ensemble Spread

Compares forecast timelines from operational global and regional NWP models:
- **ECMWF IFS** (European Centre for Medium-Range Weather Forecasts, 9km)
- **NCEP GFS** (Global Forecast System, NOAA, 13km)
- **NCMRWF NCUM** (National Centre for Medium Range Weather Forecasting, India, 12km)
- **IMD WRF** (Regional Mesoscale WRF-ARW, 3km)

Calculates inter-model standard deviation (spread) to quantify forecast certainty during convective weather.

```python
import sys
import os
import json

sys.path.append(os.path.abspath('../python'))
from analytics.model_comparison import ModelComparisonAnalytics

# 72-hour temperature forecast timelines (°C) at 6-hour intervals for New Delhi
models = {
    "ECMWF_IFS": [32.1, 35.4, 34.8, 30.2, 28.5, 33.0, 36.2, 34.0, 29.8, 28.1, 33.5, 36.8],
    "NCEP_GFS":  [32.8, 36.1, 35.5, 31.0, 29.0, 33.8, 37.1, 34.6, 30.4, 28.7, 34.2, 37.4],
    "NCMRWF":    [32.0, 35.2, 34.9, 30.0, 28.2, 33.2, 36.5, 33.9, 29.5, 27.9, 33.4, 36.6],
    "IMD_WRF":   [31.8, 35.0, 34.4, 29.8, 28.0, 32.7, 36.0, 33.5, 29.2, 27.8, 33.0, 36.2]
}

spread_analysis = ModelComparisonAnalytics.calculate_ensemble_spread(models)
print("Ensemble Spread & Confidence Analysis:")
print(json.dumps(spread_analysis, indent=2))
```

### 2. Systematic Model Bias Evaluation Against Observed Truth

```python
ground_truth = [32.0, 35.3, 34.6, 30.1, 28.4, 33.1, 36.4, 33.8, 29.6, 28.0, 33.3, 36.5]
bias_eval = ModelComparisonAnalytics.identify_model_biases(models, ground_truth)
print("Model Bias & Error Scores:")
print(json.dumps(bias_eval, indent=2))
```
